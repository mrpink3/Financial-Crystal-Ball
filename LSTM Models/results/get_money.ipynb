Non_shifted_classification

```python
import pandas as pd

finance_data = pd.read_csv("../../data/final_dataset.csv")
```

```python
stocks_in_order =['10Y Treasury','Apple','Coca-Cola','Disney','Dow Jones',
                'EUR-USD','Exxon Mobil','Fed Funds Rate','Gold','Google',
                'Intel','Johnson & Johnson','JP Morgan','Microsoft','NASDAQ Composite',
                'Oil','S&P 500','USD-JPY','Walmart']
```

```python
for stock in stocks_in_order:
    result = pd.read_csv(f"non_shifted_classification/test_predictions_Classification_{stock}.csv")
    result.rename(columns={"Dates": "Date"}, inplace=True)

    merged_data = pd.merge(result, finance_data, on="Date", how="left")
    merged_data = merged_data[["Date",stock,"Predicted"]]

    hold_money = (100 / merged_data[stock].iloc[0]) * merged_data[stock].iloc[-1]
    #print(f"Start price for {stock} is: {merged_data[stock].iloc[0]}")
    #print(f"End Price for {stock} is : {merged_data[stock].iloc[-1]}")
    print(f"Hold money for {stock} is: {hold_money}")

    money = 100
    stock_count = 0
    for index, row in merged_data.iterrows():
        if row["Predicted"] == 0:
            if stock_count != 0:
                money = stock_count * row[stock]
                stock_count = 0
        elif row["Predicted"] == 2:
            if money != 0:
                stock_count = money / row[stock]
                money = 0

    if money == 0:
        money = stock_count * merged_data[stock].iloc[-1]

    print(f"Trade money for {stock} is: {money}")
    print("---------------------------------------------")
```
